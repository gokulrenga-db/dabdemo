This default notebook is executed using Databricks Workflows as defined in resources/db_connect_example.job.yml.

```python
from dabdemo import main
import os 

print(os.getenv('WORKSPACEBUNDLEPATH'))

df = main.get_taxis(main.get_spark())
# df.show(10)
filtered_df = df.filter("pickup_zip = 10282 and fare_amount > 10")

# filtered_df.show()
dropoffzip_df = filtered_df.select("dropoff_zip").distinct()
# dropoffzip_df.show()
joined_df = df.join(dropoffzip_df, dropoffzip_df.dropoff_zip == df.pickup_zip).drop(dropoffzip_df.dropoff_zip).union(filtered_df)

joined_df.show(100)
```
